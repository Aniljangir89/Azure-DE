# How to read the data loaded in Managed Volumn

```python
df = spark.read \
    .option("header", True) \
    .option("inferSchema", True) \
    .csv("/Volumes/handson/testschema/salesdata/")
```

 Now this dataset or rdd contains lots of the data which is distributed across the workers node 
 and if we simply go to print these all values then thats not good practice because it will consume lot of time and cost 

```python
df.show(10)
```

* and if we want to see the schema of the table we can also do that

```python
df.printSchema()
```

* now see how large is this dataset --

```python
df.count()
```
