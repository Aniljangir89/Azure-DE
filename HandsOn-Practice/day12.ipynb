```python
df = spark.table("handson.testschema.messy_employee_dataset")
df.show(10)
```

```python
df.printSchema()
```

```python
df.filter(df['Employee_ID'] != 'NULL').show(10)
```

```python
# we can combine first and last name to full name by---
from pyspark.sql import functions as F
df = df.withColumn(
    "Full_Name",
    F.concat_ws(" ", F.col("First_Name"), F.col("Last_Name"))
)
```

```python
df.select(df['Full_Name']).show(10)
```

```python
df.printSchema()
```

```python
df = df.withColumn(
    "Salary",
    F.when(F.col("Salary") == "N/A", None)
     .otherwise(F.col("Salary"))
     .cast("double")
)
```

```python
df.select("Salary").show(20)
```

```python
# now lets find the employees who joins first and who joins last
df.orderBy(F.col("Join_Date")).select("Full_Name", "Join_Date").show(1)
df.orderBy(F.col("Join_Date").desc()).select("Full_Name", "Join_Date").show(1)
```

```python
# now lets find the employees who join after 2023
df.filter(F.year("Join_Date")> '2023').count()
```
