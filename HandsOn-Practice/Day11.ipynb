Join two datasets 3 different ways (inner, left, broadcast) and compare query plans with .explain().

we have one sample dataset 
```sql
handson.testschema.messy_employee_dataset
```
to perform the join operation first we can normalize this dataset into 2 datasets 

```md
INNER      → Matching rows from both
LEFT       → All rows from left
RIGHT      → All rows from right
FULL       → All rows from both
LEFT SEMI  → Matching rows from left only
LEFT ANTI  → Non-matching rows from left only
CROSS      → Every possible combination
```

````python
df = spark.table("handson.testschema.messy_employee_dataset")
````

````python
df.show(10)
````

````python
df1 = df.select("Employee_ID","First_Name","Last_Name","Salary")
df2 = df.select("Employee_ID","Department_Region","Status")
````

````python
df2.printSchema()
````

````python
# how many employees are belogning to the department 'Sales'?
from pyspark.sql.functions import col

result = df1.join(
    df2.filter(col("Department_Region") == "DevOps-New York"),
    "Employee_ID",
    "inner"
)
````

````python
df2.printSchema()
````

````python
result.show(10)
````

If you want all employees from df2 where region is DevOps-New York:

````python
result = df1.join(
    df2.filter(col("Department_Region") == "DevOps-New York"),
    "Employee_ID",
    "right"
)

result.show()
````

````python
result = df1.join(
    df2.filter(col("Department_Region") == "DevOps-New York"),
    "Employee_ID",
    "left"
)

result.show()
````

````python
df1.join(df2, "Employee_ID", "left_semi").show()
````
