# 01_Setup - Unity Catalog & Schemas
This notebook creates the Unity Catalog database and the three schemas (bronze, silver, gold) used in the Medallion Architecture.
You only need to run it once.

```python
from pyspark.sql import functions as F

CATALOG = "ecommerce_catalog"

# Create catalog if it does not exist
spark.sql(f"CREATE CATALOG IF NOT EXISTS {CATALOG}")

# Create schemas for each layer
spark.sql(f"CREATE SCHEMA IF NOT EXISTS {CATALOG}.bronze")
spark.sql(f"CREATE SCHEMA IF NOT EXISTS {CATALOG}.silver")
spark.sql(f"CREATE SCHEMA IF NOT EXISTS {CATALOG}.gold")

print("Catalog and schemas are ready.")
```
