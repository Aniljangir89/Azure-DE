groupBy, agg, sorting (orderBy), aggregation functions (sum, avg, count, min, max)

```python
df = spark.table("handson.testschema.messy_employee_dataset")
```

```python
df.show(10)
```

find the count of the employees in each region

```python
df.groupBy("Department_Region").count().show()
```

find the avg salary from each region

```python
from pyspark.sql.functions import avg, max, min
df = df.withColumn("Salary", df["Salary"].try_cast("double"))
df.groupBy("Department_Region").agg(avg(df["Salary"]), max(df["Salary"]), min(df["Salary"])).show()
```

who can we find the employee with max salary among all these employees


```python
from pyspark.sql.functions import col
df.filter(col("Salary").isNotNull()) \
  .orderBy(col("Salary").desc()) \
  .show(10)
```
