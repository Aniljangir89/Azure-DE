first check type of the performace score column 

```python
from pyspark.sql.functions import countDistinct
df.select("Performance_Score").distinct().show()

```

now we want to add new column to flag those employees which has performance score poor or average and they are working remotely

```python
from pyspark.sql.functions import col, when

df = df.withColumn(
    "status",
    when(
        (col("Remote_Work") == True) &
        (col("Performance_Score").isin("Poor", "Average")),
        "Critical"
    ).otherwise("Okay")
)
```
